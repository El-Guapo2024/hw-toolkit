writes `hw_toolkit` code → `write_kicad()` rewrites the `.kicad_sch` → a live
preview tab auto-reloads (KiCanvas renders it in-browser, instant, no
kicad-cli). Mirrors openscad-studio's loop.

1. Select the kernel → the project venv (`python3` / `hw-toolkit-venv`).
2. Run the setup cell. It prints a localhost URL — open it in VS Code's
   **Simple Browser** (Cmd-Shift-P → "Simple Browser: Show") for a live
   preview tab docked beside your code.
3. Talk to Claude Code. Every `board.write_kicad()` it runs repaints that tab
   by itself — no manual re-render, no notebook needed for the view.

The `board` is shared: what Claude edits is what the preview shows, live.
